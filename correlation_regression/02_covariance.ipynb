### 공분산
공분산이란 두 변수가 각각의 평균을 기준으로 어떤 방향으로 함께 변하는지 나타내는 값이다.

예를들어 공부시간과 시험점수라는 두 변수가 있을 때 공부시간이 증가할수록 시험점수도 같이 증가할 것이다.

이렇게 두 변수가 같은 방향으로 변하는 경향이 있으면 공분산은 양수가 된다.

반대로 공부 시간이 증가할수록 시험점수가 떨어진다면 공분산은 음수가 된다.

| 공분산 | 해석 |
| --- | --- |
| 양수 (+) | X가 평균보다 클 때 Y도 평균보다 큰 경향 |
| 음수 (-) | X가 평균보다 클 때 Y는 평균보다 작은 경향 |
| 0에 가까움 | 뚜렷한 선형 관계가 약하거나 없음 |

주의할 점은 공분산이 0이라고 해서 두 변수가 완전히 무관하다는 뜻은 아니다. 곡선 형태의 비선형 관계가 존재할 수도 있다.

```python
# 공분산 구하는 공식은 아래와 같다.
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([50, 60, 70, 80, 90])

x_mean = x.mean()
y_mean = y.mean()

# x - x_mean : x의 편차
# y - y_mean : y의 편차
# 두 편차를 곱했을 떼 두 변수의 부호가 같다면 양수 다르다면 음수가 될 것이다.
# 이 편차들의 곱을 모두 더해 공분산을 계산한다.
covariance = np.sum(
    (x - x_mean) * (y - y_mean)
) / (len(x) - 1)

print("X 평균:", x_mean)
print("Y 평균:", y_mean)
print("표본 공분산:", covariance)
```

```python
# Numpy 함수를 이용해서 간단하게 구할 수 있다.
# [[  2.5  25. ]
#  [ 25.  250. ]]
# 2.5 : x의 표본분산
# 250 : y의 표본분산
# 25 : x와 y의 표본 공분산
# 공분산 행렬은 대칭이기 때문에 [0, 1]과 [1, 0]의 값이 같다.
cov_matrix = np.cov(x, y)
print(cov_matrix)
```

공분산은 단위의 영향을 받기 때문에 단위가 달라지면 공분산의 값이 달라질 수 있다.

그렇기 때문에 공분산의 절대값이 크다고 해서 둘 사이의 관계가 더 강해졌다고 보기는 어렵다.

예를들어 키와 몸무게의 상관관계를 알기위해 공분산을 계산했을때 키를 m로 계산했을때와 cm로 계산했을 때 공분산 값은 100배 차이가 난다.

```python
# 실습1. 간단한 데이터셋을 이용해서 모집단의 공분산과 표준의 공분산 계산해보기

x = np.array([2, 4])
y = np.array([3, 5])

x_mean = x.mean()
y_mean = y.mean()

# 모집단의 공분산
population_cov = np.sum(((x - x_mean) * (y - y_mean))) / len(x)
print(population_cov)

# 표본 공분산
sample_cov = np.sum(((x - x_mean) * (y - y_mean))) / (len(x) - 1)
print(sample_cov)


# --------------------------
# numpy를 이용해서 간단히 계산

# ddof=0 모집단의 공분산
print(np.cov(x, y, ddof=0))

# ddof=1 표본 공분산
print(np.cov(x, y, ddof=1))
```

```python
# 실습2. 3개의 데이터셋을 이용해서 공분산 확인해보기

x = np.array([1, 3, 5, 7])
y = np.array([4, 7, 4, 8])
z = np.array([4, 2, 7, 3])

x_mean = x.mean()
y_mean = y.mean()
z_mean = z.mean()

# x와 y의 표본 공분산
# 3.0 : x와 y는 공분산값이 양수이므로 두 변수는 양의 상관관계를 나타낸다
x_y_sample_cov = np.sum((x - x_mean) * (y - y_mean)) / (len(x) - 1)
print(x_y_sample_cov)

# x와 z의 표본 공분산
# 0.666 : x와 z는 공분산값이 양수이므로 두 변수는 양의 상관관계를 나타낸다
x_z_sample_cov = np.sum((x - x_mean) * (z - z_mean)) / (len(x) - 1)
print(x_z_sample_cov)
```
