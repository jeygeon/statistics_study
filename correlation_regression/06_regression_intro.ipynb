### 회귀분석 기초(Regression Introduction)
상관분석은 두개 이상의 변수 사이의 관계가 얼마나 강한지, 어떤 방향으로 움직이는지 관계를 나타냈다면

회귀분석(Regression Analysis)은 하나 이상의 설명변수(X)를 이용해 목표변수(Y)의 변화를 설명하거나 예측하는 통계적 분석 방법이다.

예를들면 다음과 같다.

> 공부시간이 1시간 증가하면 시험점수는 평균적으로 몇 점 증가할까? 그리고 7시간 공부한다면 시험점수는 몇 점으로 예측할 수 있을까?

회귀분석에서는 변수의 역활을 구분해야 한다.

| 구분 | 의미 | 예시 |
| --- | --- | --- |
| 독립변수(X) | Y를 설명하거나 예측하는데 사용하는 변수 | 공부시간 |
| 종속변수(Y) | 설명하거나 예측하려는 변수 | 시험점수 |

### 단순 선형 회귀식
가장 기본적인 회귀분석은 `단순 선형 회귀`(Simple Linear Regression)이다.

단순 선형 회귀는 두 변수 사이의 관계를 직선으로 표현해서, 그 직선을 이용해 값을 예측하는 방법이다.

아래 예시 데이터로 **회귀 직선**을 구해보겠다


````python
import numpy as np
import matplotlib.pyplot as plt

# 데이터
study_hours = np.array([1, 2, 3, 4, 5, 6])
scores = np.array([50, 55, 65, 70, 80, 90])

# 1차 회귀직선 추정
# slope : 기울기
# intercept : y 절편
slope, intercept = np.polyfit(study_hours, scores, deg=1)

# 위에서 나온 회귀 직선으로 알 수 있는 내용은
# 공부를 1시간찍 늘릴수록 점수가 8점씩 오르고
# 공부시간이 0시간일 경우 40점 정도일 것이다.
print("기울기 :", slope) # 8
print("절편 :", intercept) # 40

# 예측값 계산
y_pred = slope * study_hours + intercept
print(y_pred)
````

위에서 나온 회귀직선을 통해 나온 결과는 아래와 같다.

| 공부시간 | 실제 점수 | 예측 점수 |
| --- | --- | --- |
| 1 | 50 | 48.33 |
| 2 | 55 | 56.33 |
| 3 | 65 | 64.33 |
| 4 | 70 | 72.33 |
| 5 | 80 | 80.33 |
| 6 | 90 | 88.33 |

단순 선형 회귀식에서 나온 결과가 실제 결과와의 차이가 좀 있는 것을 확인할 수 있다.

이 차이가 바로 **잔차**(Residual)라고 한다.

#### 잔차 (Residual)

잔차는 실제 관측값과 회귀모형의 예측값 사이의 차이이다.

예를들어 공부시간이 3시간인 학생의 실제 점수가 65점이고 예측 점수가 64.33 이면 잔차는 0.66이다.

```python
# 잔차 구하기
# 실제 점수 - 예측 점수
residuals = y - ypred
```
