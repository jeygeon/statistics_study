### Peason Correlation Coefficient (피어슨 상관계수)

피어슨 상관계수란 두 수치형 변수 사이의 선형 관계의 방향과 강도를 나타내는 통계량이다.

이전에 학습한 공분산은 단위에 따라서 수치가 크게 달라지기도 하고 그 값에 따라서 두 변수간 상관관계가 큰지 작은지 확인하기는 한계가 있었다.

피어슨 상관계수는 이러한 단위의 영향을 제거해서 서로 다른 변수 쌍의 선형 관계를 비교할 수 있게 해준다.

#### 피어슨 상관계수의 범위

피어슨 상관계수는 `r`로 보통 표현하고, 항상 -1부터 1까지의 범위를 가진다.



위의 그림과 같이 -1로 갈수록 음의 선형 관계를 가지고 1로 가까이 갈수록 양의 선형 관계를 가진다.

기억할 점은 두가지이다.
- **부호 (+/-)** : 두 변수가 어느 방향으로 함께 변하는가
- **절대값 (|r|)** : 선형관계가 얼마나 강한가

예를들어 피어슨 상관관계를 뜻하는 r 값이 -0.9는 +0.4보다 선형 관계가 더 강하다.

부호가 음수라는 이유로 약한 관계가 되는것은 아니다.

```python
# 피어슨 상관계수 계산
import numpy as np

x = np.array([1, 3, 5, 7])
y = np.array([4, 7, 4, 8])

# x와 y의 피어슨 상관계수
# x와 y의 공분산
x_y_smaple_cov = np.cov(x, y, ddof=1)[0, 1]

# x와 y의 표준편차
x_std = x.std(ddof=1)
y_std = y.std(ddof=1)

# 피어슨 상관계수 계산
x_y_peason_r = x_y_smaple_cov / (x_std * y_std)
print(x_y_peason_r)



# --------------------------
# numpy 라이브러리를 이용해서 쉽게 구할 수도 있다.
# corrcoef()는 상관계수 행렬을 가져오기 때문에
# 원하는 값을 가져오고 싶으면 [0, 1]을 추가한다.
print(np.corrcoef(x, y)[0, 1])
```

```python
# 실습문제 1.
# 피어슨 상관계수 직접 구해보기

import numpy as np

x = np.array([1, 3, 5, 7])
y = np.array([4, 7, 4, 8])
z = np.array([4, 2, 7, 3])

# X, Y, Z의 표본 표준편차를 각각 계산
x_std = x.std(ddof=1)
y_std = y.std(ddof=1)
z_std = z.std(ddof=1)

# X-Y의 표본 공분산을 계산
x_y_smaple_cov = np.cov(x, y, ddof=1)[0, 1]

# X-Y의 피어슨 상관계수를 공식을 이용해 직접 계산
x_y_peason_r = x_y_smaple_cov / (x_std * y_std)
print(x_y_peason_r)

# X-Z의 피어슨 상관계수도 직접 계산
x_z_smaple_cov = np.cov(x, z, ddof=1)[0, 1]
x_z_peason_r = x_z_smaple_cov / (x_std * z_std)
print(x_z_peason_r)
```
