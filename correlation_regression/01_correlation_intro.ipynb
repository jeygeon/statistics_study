### 1. 상관관계 기초 (Correlation Intro)

상관관계란 두 변수가 함께 변하는 방향과 관계의 강도를 의미한다.

상관관계에는 아래 세 가지 유형이 있다.

1. 양의 상관관계 : X가 증가할수록 Y가 증가
2. 음의 상관관계 : X가 증가할수록 Y가 감소
3. 선형 상관관계 없음 : X와 Y 사이에 뚜렷한 관계가 보이지 않음

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "study_hours": [1, 2, 3, 4, 5, 6, 7],
    "exam_score": [45, 50, 60, 65, 75, 80, 90]
}

df = pd.DataFrame(data)

plt.figure(figsize=(4, 3))

plt.scatter(df["study_hours"], df["exam_score"])

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")

plt.show()
```

위 그래프로 알 수 있는 내용은 공부시간이 늘어나면 시험 점수가 늘어난다는 사실을 알 수 있다.

여기서 한가지 포인트는 **상관관계가 있다고 해서 반드시 인과관계가 있다는 뜻은 아니라는 것**이다.

예를들어 아이스크림 판매량과 온열질환 환자 발생 건수가 같이 증가했다고 하더라도, 아이스크림이 온열질환을 유발한다고 결론을 내릴수는 없다.

기온이라는 다른 요인이 두 변수에 모두 영향을 줄 수 있기 때문이다.
