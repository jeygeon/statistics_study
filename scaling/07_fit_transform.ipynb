### 💡 fit(), transform(), fit_transform()의 이해

지금까지 스케일링을 할 때 아래와 같이 사용했다.
```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
```

#### 1️⃣ fit() 이란?
`fit()`은 데이터를 분석해서 스케일링에 필요한 기준값을 계산하고 저장하는 함수이다.

예를들어 `StandardScaler`는 스케일링을 하기위해 평균과 표준편차가 필요하다.

`fit()`함수는 이렇게 평균을 계산하고 표준편차를 계산하는 역활을 한다.

````python
# fit() 확인해보기
from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler
)
import numpy as np

X = np.array([
    [10],
    [20],
    [30],
    [40],
    [50]
])

standard_scaler = StandardScaler()
minmax_scaler = MinMaxScaler()

standard_scaled_data = standard_scaler.fit(X)
minmax_scaled_data = minmax_scaler.fit(X)

print("평균 :", standard_scaled_data.mean_)
print("표준 편치 :", standard_scaled_data.scale_)
print("최대값 :", minmax_scaled_data.data_max_)
print("최소값 :", minmax_scaled_data.data_min_)
````

위 코드에서 알 수 있는것은 `fit()`은 단순 평균 계산 표준편차 계산만 하는것이 아니라 적용하려는 스케일러에 맞게 필요한 데이터를 변환해주는 것이다.

#### 2️⃣ transform() 이란?
`transform()`은 `fit()`에서 계산한 기준값을 이용해 실제 데이터를 변환하는 함수이다.

````python
# fit()에서 저장한 평균과 표준편차를 사용
standard_scaled = standard_scaler.transform(X)

print(standard_scaled)
````

#### 3️⃣ fit_transform() 이란?
`fit_transform()`은 함수 이름 그대로 `fit()`과 `transform()`을 연속으로 실행하는 함수이다.

#### 4️⃣ 학습 데이터와 테스트 데이터란?

머신러닝에서는 일반적으로 데이터를 두 부분으로 나누어 사용한다.
- 학습 데이터(Train Data) : 모델이 학습할 때 사용하는 데이터
- 테스트 데이터(Test Data) : 학습이 끝난 모델의 성능을 평가하는 데이터

테스트 데이터에는 `transform()`만 사용하는데 그 이유가 무엇일까

아래와 같은 데이터가 있다고 해보지

````python
X_train = np.array([
    [10],
    [20],
    [30],
    [40],
    [50],
])

X_test = np.array([
    [60],
    [70],
])

scaler = StandardScaler()

# 학습 데이터로 기준값 계산 및 변환
X_train_scaled = scaler.fit_transform(X_train)

# 학습 데이터에서 계산한 기준값을 그대로 사용
X_test_scaled = scaler.transform(X_test)

print("학습 데이터:")
print(X_train_scaled)

print("테스트 데이터:")
print(X_test_scaled)
````

StandardScaler를 적용한다고 해서 테스트 데이터의 평균까지 0이 되는것은 아니다.

학습 데이터의 평균과 표준편차를 기준으로 변환하기 때문에 테스트 데이터의 값이 2나 3보다 커질 수 있다.

그렇다고 테스트 데이터에 다시 `fit_transform()`을 하게되면 테스트 데이터에 맞게 새롭게 평균이나 표준편차를 계산하기 때문에

학습한 데이터의 값을 무시해버려서 학습 데이터의 존재가 의미가 없게 된다.
