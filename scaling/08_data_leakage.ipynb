### Data Leakage (데이터 누수)

데이터 누수란 실제 예측 사용할 수 없는 정보나 평가 데이터 정보가 학습 또는 전처리에 사용되는 것이다.

예를들어 어떤 회사에서 직원의 연봉을 예측하는 머신러닝 모델을 만든다고 해보자.

학습 데이터는 다음과 같다.
| 직원 | 경력 | 나이 | 연봉 |
| --- | --- | --- | --- |
| A | 1년 | 25 | 3,000 |
| B | 3년 | 28 | 4,000 |
| C | 5년 | 32 | 5,000 |
| D | 7년 | 35 | 6,000 |

여기서 모델은 경력과 나이를 이용해 연봉을 예측해야 한다.

그런데 테스트 과정에서 정답 연볻을 미리 사용한다면, 테스트 성능을 신뢰하기 어려워진다.

이처럼 데이터 누수는 모델의 성능을 실제보다 좋게 보이게 만들거나, 평가의 신뢰성을 떨어뜨릴 수 있다.

```python
X_train = np.array([
    [10],
    [20],
    [30],
    [40],
    [50]
])

X_test = np.array([
    [60],
    [70]
])
```

이러한 데이터가 있을 때 학습 데이터의 평균은 30이다. 하지만 학습 데이터와 테스트 데이터의 평균은 40이다.

만약 전체 데이터로 스케일링을 진행했다면 테스트 데이터의 값 60, 70이 이미 평균 계산에 사용된 것이다.

즉, **테스트 데이터의 정보가 학습 전처리 과정에 들어간 것**이다.

````python
# 올바른 방법은
# 데이터 분리 -> 학습 데이터로 fit -> 테스트 데이터로 transform 순서이다.

import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

X = np.array([
    [10], [20], [30], [40], [50], [60], [70]
])

# 1. 먼저 데이터 분리
X_train, X_test = train_test_split(
    X,
    test_size=0.3,
    random_state=42
)

# 2. 스케일러 생성
scaler = StandardScaler()

# 3. 학습 데이터로만 fit
X_train_scaled = scaler.fit_transform(X_train)

# 4. 테스트 데이터는 transform만
X_test_scaled = scaler.transform(X_test)
````
