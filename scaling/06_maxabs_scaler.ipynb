### 💡 MaxAbsScaler (절대 최댓값 스케일링)

#### 1️⃣ MaxAbsScaler란
MaxAbsScaler란 각 열에서 절댓값이 큰 값을 찾아, 모든 데이터를 그 값으로 나누는 스케일링 방법이다.

MaxAbs는 Maximum Absolute Value, 즉 절대 최대값을 의미한다.

예를들어 아래와 같은 데이터가 있다고 했을 떄
```python
scores = [-10, 5, 15]
```

데이터중 절대값이 가장 큰 값은 15이다. 따라서 모든 값을 15로 나눈다고 했을 때 변환된 결과는 다음과 같다.

`[-0.667, 0.333, 1]`

**MaxAbsScaler**의 중요한 특징은 다음과 같다.
- 학습 데이터의 변환 결과가 -1 ~ 1 범위에 들어간다.
- 원본 데이터의 양수와 음수 부호를 유지한다.
- 원래 값이 0이면 변환 후에도 0이다.
- 평균이나 중앙값을 빼지 않는다.

````python
# Numpy로 구현해보기
import numpy as np

scores = np.array([-10, 5, 15])

# 절대값 계산
abs_scores = np.abs(scores)

# 절대 최댓값
abs_max = np.max(abs_scores)

# MaxAbs Scaling
scaled_scores = scores / abs_max

print("원본 데이터:", scores)
print("절댓값:", abs_scores)
print("변환 결과:", scaled_scores)
````

````python
# scikit-learn의 MaxAbsScaler 사용해보기
from sklearn.preprocessing import MaxAbsScaler

scores = np.array([-10, 5, 15])

# 1차원 배열을 2차원 배열로 변환
X = scores.reshape(-1, 1)

# MaxAbsScaler 객체 생성
scaler = MaxAbsScaler()

# 스케일링
X_scaled = scaler.fit_transform(X)

print("원본 데이터:")
print(X)

print("변환 결과:")
print(X_scaled)
````
