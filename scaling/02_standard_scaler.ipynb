### 표준화 (StandardScaler)

표준화는 스케일링 방법 중 하나이다. 표준화는 데이터의 평균을 0, 표준편차를 1로 변환하는 스케일링 방법이다.
- 평균 (Mean) : 데이터의 중심을 나타내는 값
- 표준편차 (Standard Deviation) : 데이터가 평균을 중심으로 얼마나 퍼져있는지 나타내는 값

데이터를 표준화하는 방법은 표준화하려는 데이터에 평균을 빼고 표준 편차로 나누면 된다.

파이썬 코드로 직접 계산해보자면 아래와 같다.

```python
# 파이썬 코드로 데이터 스케일링하기 (표준화)
import numpy as np

scores = np.array([60, 70, 80, 90, 100])

scores_mean = scores.mean() # 80
scores_std = scores.std() # 14.1421

# 표준화한 원래 데이터
standard_scaling = (scores - scores_mean) / scores_std
print(standard_scaling)
```

위의 표준화 방법으로 원래 데이터가 아래와 같이 스케일링 되었다.

| 표준화 전 데이터 | 표준화 후 데이터 |
| --- | --- |
| 60 | -1.414 |
| 70 | -0.707 |
| 80 | 0 |
| 90 | 0.707 |
| 100 | 1.414 |


```python
# scikit-learn 라이브러리의 StandardScaler 사용해보기
import numpy as np
from sklearn.preprocessing import StandardScaler

scores = np.array([60, 70, 80, 90, 100])

# StandardScaler는 2차원 데이터를 입력받아야 해서
# 1차원 데이터인 scores를 2차원으로 변환해준다.
# reshape(행 갯수, 열 갯수) : -1은 데이터 갯수에 맞춰서 행을 알아서 생성 
# -1은 원소 수에 맞게 행 갯수를 자동으로 계산 : 여기서는 5행
X = scores.reshape(-1, 1)

scaler = StandardScaler()

# fit_transform() : fit과 transform 두 작업을 한꺼번에 실행하는 함수
# fit() : 데이터의 평균과 표준편차를 계산해 변환 기준을 학습
# transform() : 학습한 평균과 표준편차를 이용해 실제 결과를 반환
X_scaled = scaler.fit_transform(X)

print("원본:")
print(X)

print("표준화 결과:")
print(X_scaled)
```

```python
# 여러 열을 동시에 표준화하기
import pandas as pd
from sklearn.preprocessing import StandardScaler

data = {
    "age": [20, 30, 40, 50, 60],
    "salary": [3000, 4000, 5000, 6000, 7000]
}

df = pd.DataFrame(data)

# StandardScaler 객체 생성
scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)
print(scaled_df)
```
