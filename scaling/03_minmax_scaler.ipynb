### Min-Max Scaling
Min-Max Scaling은 데이터를 최소값과 최대값을 기준으로 일정한 범위로 변환하는 스케일링이다.

기본적으로 데이터를 0~1사이로 변환한다.

예를들어 아래와 같은 시험점수가 있다고한다면
```python
scores = [20, 40, 60, 80, 100]
```

다음과 같이 변환할 수 있다.
| 원본 점수 | 변환된 값 |
| --- | --- |
| 20 | 0 |
| 40 | 0.25 |
| 60 | 0.5 |
| 80 | 0.75 |
| 100 | 1 |


> 값의 범위는 바뀌지만, 값의 순서와 간격의 비율은 일정하게 유지한다.

여기서 포인트는 원본 데이터의 순서가 유지된다는 점이다.

````python
# min-max scaler 구현해보기
import numpy as np

scores = np.array([20, 40, 60, 80, 100])

min_value = scores.min()
max_value = scores.max()

scaled_scores = (scores - min_value) / (max_value - min_value)

print("원본 데이터 :", scores)
print("스케일링 데이터 :", scaled_scores)
````

````python
# scikit-learn의 MinMaxScaler를 사용할 수 있다.
import numpy as np
from sklearn.preprocessing import MinMaxScaler

scores = np.array([20, 40, 60, 80, 100])

# 1차원 배열을 2차원 배열로 변환
X = scores.reshape(-1, 1)

# MinMaxScaler 객체 생성
scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)

print("원본 데이터 :", X)
print("스케일링 데이터 :", X_scaled)
````

````python
# 여러 열을 동시에 스케일링 하기
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {
    "age": [20, 30, 40, 50, 60],
    "salary": [3000, 4000, 5000, 6000, 7000]
}

# 데이터를 데이터프레임으로 변환
df = pd.DataFrame(data)

# 스케일링 객체 생성
scaler = MinMaxScaler()

# 데이터프레임 원본 데이터를 스케일링 진행
scaled_data = scaler.fit_transform(df)

# 스케일링된 데이터를 다시 데이터 프레임으로 변환
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
print(scaled_df)
````

min-max scaling은 이상치에 매우 취약하다.

예를들어 아래와 같은 데이터가 있다고 가정해보자

```python
scores = [10, 20, 30, 40, 1000]
```

해당 값들을 스케일링 했을때는 아래와 같은 값이 나오게 된다.

```python
[0, 0.0101, 0.0202, 0.0303, 1]
```

이렇게 되면 데이터들 사이의 차이를 구분하기 어려워질 수 있다.

min-max scaler의 주의할 점이 한가지가 더 있는데, 나중에 새롭게 들어온 데이터가 최소값보다 작거나 최대값보다 큰 값을 가지게 된다면

0~1 사이를 벗어날수도 있다는 점이다.
