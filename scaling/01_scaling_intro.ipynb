### 1. 스케일링 (Scaling)
스케일링은 데이터의 수치적 크기를 일정한 기준에 맞게 조정하는 과정이다.

수치적 크기를 조정하다는 의미로 아래 예시를 들면

| 학생 | 키(cm) | 몸무게(kg) | 연봉(원) |
| --- | --- | --- | --- |
| A | 170 | 65 | 30,000,000 |
| B | 180 | 75 | 50,000,000 |
| C | 160 | 55 | 40,000,000 |

이런 데이터를 그대로 사용하면 일부 분석 방법에는 연봉처럼 숫자의 크기가 큰 변수가 결과를 지나치게 좌우할 수 있다.

스케일링은 이렇게 다른 숫자 범위를 조정해주는 과정이다.

- 전처리(Preprocessing) : 데이터를 분석하기 좋은 상태로 가공하는 작업
- 특성(Feature) : 데이터에서 분석에 사용하는 각각의 변수 또는 열

### 2. 스케일링이 필요한 이유

#### 2-1. 서로 다른 척도를 비교
예를들어 학생을 평가할 때 점수와 학점을 사용해본다고 가정해보자

단순 점수와 학점을 그대로 더하면 점수의 영향이 훨씬 커진다.

하지만 점수가 학점보다 반드시 더 중요한 것은 아니다.

그래서 변수간의 서로 다른 척도를 적절히 조절할 필요가 있다.

#### 2-2. 거리 계산에 영향을 주기 때문
일부 머신러닝 알고리즘은 데이터 사이의 거리를 계산해서 비슷한 데이터를 찾는다.

예를들어 나이랑 연봉이 있는 데이터셋이 있고 이 데이터셋을 그대로 거리 계산에 사용한다면,

연봉의 수치가 나이에 비해 너무 크기 때문에 나이는 고려되지않고 연봉만으로 결과를 지배할 수 있게 된다.

예를들면 아래와 같은 데이터가 있고 사용자 A와 가장 비슷한 사람을 찾으려고 한다고 했을 때
| 사람 | 나이 | 연봉 |
| --- | --- | --- |
| A (기준) | 25세 | 5,000 만원 |
| B | 26세 | 6,000 만원 |
| C | 55세 | 5,000 만원 |

- B : 나이는 1살 차이지만, 연봉은 1천만원 차이
- C : 연봉은 같이만, 나이가 30살 차이

뭔가 우리의 감으로 생각했을 때 A와 가장 비슷한 사람은 B라는 것을 감각적으로 알 수 있다.

하지만 두 데이터 사이의 거리를 구할 때 자주 사용하는 유클리드 거리(Euclidean Distance)를 사용해 보았을때,

**A와 B 사이의 거리 : 약 10,000,000<br/>**
**A와 C 사이의 거리 : 30**

이라는 결과가 나오게 된다.

계산 결과에 따르면 C가 A와 압도적으로 가까운 사람이 되어버린다.

이렇듯 숫자의 크기가 크다는 이유로 해당 변수가 더 중요하게 취급되는 문제를 줄이는 것이 스케일링의 목적이다.
