### Robust Scaler
robust scaler란 데이터의 중앙값곽 사분위 범위를 이용해 데이터를 변환하는 스케일링 기법이다.

기본적으로 중앙값(Median)을 0으로 변환, 사분위(IQR)을 1로 변환한다.

앞에서 학습한 `StandardScaler`나 `MinMaxScaler`는 모두 이상치에 영향을 크게 받는 스케일링 방법이지만

`Robust Scaler`는 이상치에 영향을 덜 받는 스케일링 방법이다.

중앙값을 사용하는 이유는 이상치에 영향을 덜 받기 위해서이다.

예를들어 아래와 같은 월급 데이터가 있다고 했을 때

```python
salary = [200, 250, 300, 350, 400]
```
평균은 300이고, 중앙값도 300이다.

하지만 이상치가 아래와 같이 있는 경우에는

```python
salary = [200, 250, 300, 350, 5000]
```

평균은 1220이고, 중앙값은 300이 된다.

중앙값을 사용하면 가운데 값을 기준으로 하기 때문에 이상치에 덜 민감하게 반응한다.

그리고 IQR도 가운데 데이터 50%정도가 얼마나 퍼져있냐를 확인하기 때문에 최대값 최소값의 영향을 덜 받아서 상대적으로 안정적인 기준이 될 수 있다.

````python
# numpy로 직접 구현해보기
import numpy as np

scores = np.array([10, 20, 30, 40, 50])

# 중앙값
scores_median = np.median(scores)

# 제1사분위수
q1 = np.percentile(scores, 25)

# 제3사분위수
q3 = np.percentile(scores, 75)

# IQR
iqr = q3 - q1

# 스케일링
scaled_scores = (scores - scores_median) / iqr
print("변환 결과:", scaled_scores)
````

````python
# scikit-learn의 RobustScaler를 사용해서 간단하게 스케일링도 가능하다.
import numpy as np
from sklearn.preprocessing import RobustScaler

scores = np.array([10, 20, 30, 40, 50])

# 1차월 배열을 2차원으로 변환
X = scores.reshape(-1, 1)

# RobustScaler 객체 생성
scaler = RobustScaler()

# 스케일링
X_scaled = scaler.fit_transform(X)

print("원본 데이터:")
print(X)

print("변환 결과:")
print(X_scaled)
````

````python
# 만약 이상치가 있다면 스케일링 데이터는 어떻게 남을까
import numpy as np
from sklearn.preprocessing import RobustScaler

scores = np.array([10, 20, 30, 40, 1000])

X = scores.reshape(-1, 1)

scaler = RobustScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled)
````

RobustScaler는 이상치에 영향을 덜 받는 중앙값과 IQR을 이용해서 계산하기 때문에 극단적으로 큰 값은 스케일링 이후에도 크게 남아있을 수 있다.


````python
# 여러 변수를 동시에 스케일링 하기
import pandas as pd
from sklearn.preprocessing import RobustScaler

data = {
    "age": [20, 30, 40, 50, 60],
    "salary": [3000, 4000, 5000, 6000, 50000]
}

df = pd.DataFrame(data)

scaler = RobustScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print(scaled_df)
````

RobustScaler는 이상치를 그대로 두고 스케일링 하기 때문에 값이 -1~1 사이로 정해지진 않는다.
