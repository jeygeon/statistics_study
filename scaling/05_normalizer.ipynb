### 💡 Normalizse : 벡터 정규화
`Normalizer`란 앞에서 다뤘던 세가지 스케일링과는 좀 다르게 각 행(row)를 하나의 벡터로 보고, 그 벡터의 길이가 1이 되도록 변환하는 방법이다.

#### 1️⃣ 벡터(Vector)란?
벡터는 크기와 방향을 가진 양이다. 

예를들어 아래와 같은 데이터가 있다고 해보면
```python
vector = np.array([3, 4])
```

이 벡터는 2차원 좌표 평면에서 원점 (0, 0)부터 (3, 4)를 향하는 화살표라고 생각할 수 있다.

데이터 분석에서는 다음처럼 한 사람의 여러 특성을 벡터로 표현할 수 있다.

```python
# 한 사람의 데이터를 세 개의 숫자로 표현 (나이, 키, 몸무게)
person = np.array([25, 175, 70])
```

### 2️⃣ 벡터의 차원(Dimension)
벡터가 가진 성분의 갯수를 차원이라고 한다.

```python
# 성분이 2개라 2차원 벡터
vector_2d = np.array([3, 4])

# 성분이 3개라 3차원 벡터
vector_3d = np.array([3, 4, 5])
```

벡터의 차원은 numpy 배열의 차원과는 좀 다른점이

numpy에서는 `np.array([3, 4])`가 성분이 2개인 1차원 배열이지만,

벡터에서는 성분이 2개인 2차원 배열이 된다.

````python
# 벡터의 길이 계산하기
import numpy as np

vector = np.array([3, 4])

length = np.sqrt(np.sum(vector ** 2))

print("벡터의 길이 :", length)
````

````python
# numpy에는 벡터의 길이를 계산하는 함수도 있다.
# np.linalg.norm() : 1차원 벡터에 대해 길이를 계산한다.
length = np.linalg.norm(vector)
print(length)
````

**벡터 정규화(Vector Normalization)** 는 벡터의 방향을 유지하면서 길이를 1로 만드는 과정이다.

길이가 1인 벡터를 단위벡터(Unit Vector)라고 부른다.

벡터 정규화를 하는 방법은 각 성분을 전체의 길이로 나누면 된다.

벡터 정규화를 하게되면 원본 벡터 성분이 변경되더라도 정규화 후의 길이는 항상 1이 된다.

````python
# 벡터 정규화해보기
vector = np.array([3, 4])

# 벡터 길이 계산
length = np.linalg.norm(vector)

# 벡터 정규화
# 벡터 정규화는 벡터의 모든 성분을 길이로 나눈다.
normalized_vector = vector / length

print("정규화된 벡터 :", normalized_vector)
print("정규화된 벡터의 길이 :", np.linalg.norm(normalized_vector))
````

````python
# scikit-learn의 Normalizer 사용해서 정규화하기
from sklearn.preprocessing import Normalizer

X = np.array([
    [3, 4],
    [6, 8]
])

# Normalizer 객체 생성
# norm="l2" : 원점에서 벡터 끝점까지의 거리, 즉 벡터의 길이
normalizer = Normalizer(norm="l2")

# 정규화 적용
X_normalized = normalizer.transform(X)

# 정규화된 벡터의 길이
lengths = np.linalg.norm(X_normalized, axis=1)

print("원본 데이터:")
print(X)

print("정규화 결과:")
print(X_normalized)

print("각 행의 L2길이 확인:")
print(lengths)
````

벡터 정규화는 보통 텍스트 데이터 분석에 자주 사용된다.

에를들어 두 문서에 등장하는 단어의 빈도를 분석한다고 해볼때

| 문서 | Python | Java |
| --- | --- | --- |
| 문서 A | 3 | 4 |
| 문서 B | 6 | 8 |

문서 B는 두 단어가 등장하는 횟수가 문서 A보다 2배 많다.

하지만 두 단어의 등장 비율은 같다.

이떄 정규화를 적용하게 되면 두 문서 모두 `[0.6, 0.8]`이 된다.

이처럼 단어가 등장하는 상대적인 패턴을 비교할 때 활용할 수 있다.
