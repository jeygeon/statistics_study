### Z-Score (표준점수)

`Z-Score`는 특정 데이터가 평균에서 표준편차의 몇배만큼 떨어져 있는지 나타내는 값이다.

```python
import numpy as np

scores = np.array([50, 60, 70, 80, 90])

mean = scores.mean()
std = scores.std()

z_scores = (scores - mean) / std

print("평균:", mean)
print("표준편차:", std)
print("Z-Score:", z_scores)
```

z-score가 유용한 이유는 서로 다른 두 데이터를 비교할 때 유용하다.

예를들어 서로 다른 시험을 본 두 학생의 시험 결과가 아래와 같다고 가정했을 때

| 구분 | A 학생 | B 학생 |
| --- | --- | --- |
| 시험 점수 | 90점 | 80점 |
| 시험 평균 | 70점 | 60점 |
| 표준 편차 | 10점 | 5점 |
| Z-Score | 2 | 4 |

A 학생은 평균보다 표준편차 2개만큼 높고, B 학생은 평균보다 표준편차 4개만큼 높다.

따라서 원점수는 A가 높지만, **자신이 속한 시험에서 상대적인 위치는 B가 더 높다고 해석할 수 있다.**

```python
# 타이타닉 나이 데이터셋으로 z-score 계산해보기
import pandas as pd
import numpy as np

df = pd.read_csv("./datasets/titanic.csv")

# 평균과 표준편차 계산
mean_age = df["age"].mean()
std_age = df["age"].std()

# z-score 열 생성
df["age_z_score"] = (df["age"] - mean_age) / std_age
print(df[["age", "age_z_score"]].head(3))

print("Z-Score 평균:", df["age_z_score"].mean())
print("Z-Score 표준편차:", df["age_z_score"].std())
```

```python
# z-score로는 이상치를 탐색하는데 활용될 수 있다.
# 정규분포라면 약 99.7%가 평균에서 표준편차 3개 이내에 있다.
# 표준편차에 절대값 씌운 값이 3을 넘어간다면 이상치로 특정할 수 있다.
# abs() : 절대값 계산
outliers = df[df["age_z_score"].abs() > 3]
print(outliers[["age", "age_z_score"]])
```
