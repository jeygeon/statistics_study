### 분산과 표준편차

```python
import numpy as np

group_a = np.array([68, 69, 70, 71, 72])
group_b = np.array([30, 50, 70, 90, 110])

print("A 평균:", group_a.mean())
print("B 평균:", group_b.mean())
```

위의 두 그룹 A와 B는 둘다 평균이 70이다. 하지만 A는 전체적으로 평균 근처에 점수가 분포되어있고, B는 상대적으로 넓게 점수가 퍼져있다.

여기서 알 수 있는 것은 평균만으로 데이터의 퍼짐정도는 알 수 없다는 것이다.

데이터의 퍼짐정도를 수치화하기 위해서는 **분산**과 **표준편차**를 사용한다.

#### 1. 편차 (Deviation)
편차는 개별 데이터가 평균으로부터 얼마나 떨어져 있는지를 나타낸다.

```python
scores = np.array([3500, 2100, 5000, 4500])
mean = scores.mean()

deviations = scores - mean

print("평균 :", mean)
print("편치 :", deviations)
print("편차의 합 :", deviations.sum())
```

평균을 기준으로 계산한 편차의 합은 항상 0이다. 따라서 편차를 그대로 내면 데이터가 얼마나 퍼져있는지 알 수 없다.

이 문제를 해결하기 위해서 각 편차에 제곱하는 방법을 사용한다.

#### 2. 분산 (Variance)

분산은 평균에서 멀리 떨어진 거리의 제곱을 평균 낸 값이고, 파이썬 코드로 나타내면 아래와 같다.

```python
scores = np.array([3500, 2100, 5000, 4500])

mean = scores.mean()
squared_deviations = (scores - mean) ** 2

population_variance = squared_deviations.sum() / len(scores)

print("모분산:", population_variance)
print("NumPy 모분산:", np.var(scores))
```
